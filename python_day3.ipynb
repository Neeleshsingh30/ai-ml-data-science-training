<!-- Recursion -->

```python
#syntax of recursion

def recursive_function(parameters):
    if base_condition:   #base case
        return result
    else:
        return recursion_function(modified parameters)    #recursive case

    
```

```python
#Ex1: Factorial using recursion
n = int(input("enter your number"))
def factorial(n):
    if n == 1:    #base case
        return 1
    return n * factorial(n-1)    #Recursive call

print(factorial(n))
```

```python
#ex2 Fibonacci series using recursion
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n-1) + fibonacci(n-2)   #recursive call

print(fibonacci(6))
```

```python
#ex3 sum of first n natural numbers
def sum_natural(n):
    if n == 1: #base condition
        return 1
    return n + sum_natural(n-1)

print(sum_natural(5))
```

Anonymous/Lambda function

```python
#syntax of lambda function

lambda arguments: expression
```

```python
#ex1 basic lambda function
add = lambda a, b : a +b

print(add(5,8))
```

```python
multiply = lambda x,y,z : x*y*z

print(multiply(1,3,8))
```

```python
#example2:using lambda with map
numbers=[1,2,3,4,5]

doubled= list(map(lambda x:x * 2, numbers))

print(doubled)
```

```python
numbers = []

n = int(input("How many elements? "))

for i in range(n):
    value = int(input("Enter element: "))
    numbers.append(value)

print("Your List:", numbers)
```

```python
#using lambda with map()
numbers = [1,2,3,4,5,6]

double  = list(map(lambda x : x* 2, numbers))

print(double)
```

```python
#using lambda with filter()
numbers = [1,2,3,4,5,6]

evens = list(map(lambda x : x % 2 == 0, numbers))

print(evens)
```

```python
#using lambda with sorted()

students = [("Neelesh",22), ("ram" , 20), ("charlie", 23)]

sorted_students = sorted(students, key=lambda x : x [1])

print(sorted_students)
```

```python
#using lambda in reduce()
from functools import reduce

numbers = [5,6,7,8,9,10]

product = reduce(lambda x, y: x * y, numbers)

print(product)  
```

Global key word

```python
#without global (local variable)

x = 10

def my_function():
    x = 5
    print("Inside function :", x)

my_function()
print("Outside function :", x)
```

```python
#using global to modify a global varible

x = 50

def my_function():
    global x
    x = 25
    print("Inside function :", x)

my_function()
print("Outside function :", x)
```

```python
#using global with multiple variables
a = 100
b = 50

def update_value():
    global a, b
    a += 10
    b -= 20

update_value()
print(a,b)
```

```python
#global inside nestedd functions

def outer():
    global x
    x = "Hello"

    def inner():
        global x 
        x = "World"

    inner()
    print("Inside outer :", x)

outer()
print("outside function:",x)
```
